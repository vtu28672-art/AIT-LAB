```python
from collections import deque
graph={
    'A':['B','C'],
    'B':[],
    'C':[]
}
visited=set()
queue=deque()
start='A'
queue.append(start)
print("BFS Traversal:")
while queue:
  node=queue.popleft()
  if node not in visited:
    print(node,end="")
    visited.add(node)
    for neighbor in graph.get(node,[]):
      if neighbor not in visited:
        queue.append(neighbor)
```

```python
graph = {
    'A':['B','C'],
    'B':[],
    'C':[]
}
visited = set()

def dfs(node):
  if node not in visited:
    print(node, end="")
    visited.add(node)
    for neighbor in graph.get(node,[]):
      dfs(neighbor)

print("DFS Traversal:")
dfs('A')
```
