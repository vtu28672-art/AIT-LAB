```python
def is_safe(vertex, colour, graph, colors):
  for i in range(len(graph)):
    if graph[vertex][i] == 1 and colors[i] == colour:
      return False
  return True
def graph_coloring(vertex, m, graph, colors):
  if vertex == len(graph):
    return True
  for color_attempt in range(1, m + 1):
    if is_safe(vertex, color_attempt, graph, colors):
      colors[vertex] = color_attempt
      if graph_coloring(vertex + 1, m, graph, colors):
        return True
      colors[vertex] = 0
  return False

graph=[
    [0,1,1,0,0],#A
    [1,0,0,1,0],#B
    [1,0,0,1,0],#C
    [0,1,1,0,1],#D
    [0,0,0,1,0],#E
]
n = len(graph)
min_colors_found = False
for m_val in range(1, n + 1):
  colors = [0] * n
  if graph_coloring(0, m_val, graph, colors):
    print("Minimum number of colors:", m_val)
    print("color assignment:")
    for i in range(n):
      print(chr(65 + i), "-> color", colors[i])
    break
```
