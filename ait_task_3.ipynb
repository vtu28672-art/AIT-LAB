The original code contained multiple syntax errors and logical issues, making it unfeasible to correct incrementally. Here's a revised implementation of the A* algorithm for pathfinding on a grid, assuming '0' are walkable cells and '1' are obstacles.

```python
import heapq

def heuristic(a, b):
    """Calculates the Manhattan distance heuristic between two points a and b."""
    return abs(a[0] - b[0]) + abs(a[1] - b[1])

def find_path_astar(grid, start, goal):
    """
    Finds the shortest path from start to goal in a grid using the A* algorithm.
    Grid values: 0 for walkable, 1 for obstacle.
    """
    rows, cols = len(grid), len(grid[0])

    # Priority queue: (f_cost, g_cost, current_node, path_list)
    # f_cost = g_cost + h_cost
    # g_cost = cost from start to current_node
    # path_list = list of nodes from start to current_node
    pq = [(heuristic(start, goal), 0, start, [start])]

    # Keep track of the minimum g_cost to reach a node
    g_costs = {start: 0}

    while pq:
        f_cost, g_cost, current_node, path = heapq.heappop(pq)

        # If we reached the goal, return the path
        if current_node == goal:
            return path, g_cost

        # Explore neighbors
        for dr, dc in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
            neighbor_row, neighbor_col = current_node[0] + dr, current_node[1] + dc
            neighbor = (neighbor_row, neighbor_col)

            # Check boundaries and obstacles
            if 0 <= neighbor_row < rows and 0 <= neighbor_col < cols and grid[neighbor_row][neighbor_col] == 0:
                new_g_cost = g_cost + 1  # Each step costs 1

                # If this path to neighbor is better, update it
                if neighbor not in g_costs or new_g_cost < g_costs[neighbor]:
                    g_costs[neighbor] = new_g_cost
                    new_f_cost = new_g_cost + heuristic(neighbor, goal)
                    new_path = path + [neighbor]
                    heapq.heappush(pq, (new_f_cost, new_g_cost, neighbor, new_path))

    return None, None # No path found
```

### Example Usage

```python
grid = [
    [0, 0, 0],
    [0, 1, 0],
    [0, 0, 0]
]
start_node = (0, 0)
goal_node = (2, 2)

print("Grid:")
for row in grid:
    print(row)
print(f"Start: {start_node}")
print(f"Goal: {goal_node}")

optimal_path, total_cost = find_path_astar(grid, start_node, goal_node)

if optimal_path:
    print("\nOptimal path:", optimal_path)
    print("Total cost:", total_cost)
else:
    print("\nNo path found.")
```
